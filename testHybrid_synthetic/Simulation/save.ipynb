```python
import glob
import os
from PIL import Image

def create_gif(input_folder, output_gif_path, duration_ms=200):
    # Find all PNG files in the specified folder and sort them alphabetically
    search_path = os.path.join(input_folder, "*.png")
    png_files = sorted(glob.glob(search_path))
    
    if not png_files:
        print("No PNG files found in the directory.")
        return

    # Open all images and store them in a list
    frames = [Image.open(image) for image in png_files]

    # Save as an animated GIF
    # duration is in milliseconds (e.g., 300ms = 0.3 seconds per frame)
    # loop=0 means the GIF will loop infinitely
    frames[0].save(
        output_gif_path,
        format="GIF",
        append_images=frames[1:],
        save_all=True,
        duration=duration_ms,
        loop=0
    )
    print(f"GIF successfully created at: {output_gif_path}")


png_path = "/Users/svaren/src-cc3d/new_102026/"
output = "/Users/svaren/Downloads/new_102026.gif"
create_gif(png_path, output)
```
